```python
routes=[
    {"id": "R101", "from": "karachi", "to": "Bahawalpur", "time": "8 am", "price": 2500, "seats": 40},
    {"id": "R102", "from": "karachi", "to": "Islamabad", "time": "7 pm", "price": 5000, "seats": 41},
    {"id": "R103", "from": "Bahawalpur", "to": "khanpur", "time": "9 pm", "price": 1000, "seats": 42},
    {"id": "R104", "from": "Islamabad", "to": "layyah", "time": "12 pm", "price": 1500, "seats": 43},
    {"id": "R105", "from": "layyah", "to": "khanpur", "time": "3 pm", "price": 1300, "seats": 44}
]
bookings = []
print("WELCOME TO TAHA'S AND SONS BUSS TICKET BOOKING SYSTEM")

while True:
    print("\n1. view Routes")
    print("2. Search Routes")
    print("3. Book Ticket")
    print("4. View Bookings")
    print("5. Cancel Bookings")
    print("6. Generate reciept")
    print("7. exist")
    
    choice = input("Enter Your Choice")
    if choice == "1":
        print("\nAvalable Routes")
        for r in routes:
            print(r["id"], r["from"], "to", r["to"], "| Time:", r["time"], "| Price:", r["price"], "| Seats:", r["seats"])
    
    elif choice == "2":
        query = input("Enter City or ROute ID: ").lower()
        for r in routes:
            if query in r["id"].lower() or query in r["from"].lower() or query in r["to"].lower():
                print("Found:", r["id"], r["from"], "to", r["to"], "| Time:", r["time"], "| Price:", ["price"], "| Seats:", r["seats"])
    elif choice == "3":
        name = input("Enter your name: ")
        age = input("Enter your Age: ")
        r_id = input("Enter Your ID: ").upper()
    
        slected == None
        for r in routes:
            if r["id"] == r_id:
              selected = r
              break
        if selected == None:
            print("Routes Not Found")
        else:
            seats=int(input("ENter Seats To Book : "))
            if seats>0 and seats<= selected["seats"]:
                    selected["seats"] -= seats
                    b_id = "B" + str(len(bookings) +1)
                    total = seats*selected["price"]
    
                    booking = {
                         "id": b_id,
                         "name": name,
                         "route": selected["id"],
                         "seats": seats,
                         "total": total,
                         "status": "confirmed!"
                    }
                    bookings.append(booking)
                    print("Booking succesfull! Booking ID:", b_id, "total price", total)
            else:
                print("Invalid Seat")
    
    elif choice == "4":
        print("\nYour Bookings: ")
        if len(booking) == 0:
            print("No Booking Yet. ")
        else:
            for b in booking:
                print("ID:", b["id"], "| Name:", b["name"], "| Route:", b["route"], "| Seats:", b["seats"], "| Total:", b["total"], "| Status:", b["status"])
    elif choice == "5":
        b_id = input("Enter Booking Id To cancel: ")
        found_b = None
        for b in bookings:
            if b["id"] == b_id:
                found_b = b
                break
    
        if found_b == None:
            print("Booking ID Not Found")
        elif found_b["status"] == "cancelled":
            print("Already Cancelled")
        else:
            found_b["status"] = "cancelled"
            for r in routes:
                if r["id"] == found_b["routes"]:
                    r["seats"] += found_b["seats"]
                    break
            print("Booking Cancelled and seats Restored.")
    
    elif choice == "6":
        print("\nReports: ")
        total_b = len(bookings)
        conf_b = 0
        canc_b = 0
        rev = 0
        for b in bookings:
            if b["status"] == "Confirmed":
                conf_b += 1
                rev += b["total"]
            else:
                canc_b += 1
    
        print("total Booking: ",total_b)
        print("Confirmed Booking: ",conf_b)
        print("Cancelled Booking: ",canc_b)
        print("total revenue ",rev)
    
    elif choice == "7":
        print("Exit")
        break
    else:
        print("Please enter VAlid Option")
```
